## Prepare Data for Prophet and Split

### Subtask:
Prepare the DataFrame in Prophet's required `ds` and `y` format. Then, split the data into a training set (January 2020 - December 2023) and a testing set (January 2024 - December 2024).


**Reasoning**:
The subtask requires preparing the data for Prophet by renaming columns and splitting the data into training and testing sets based on specified date ranges. This code block will handle the renaming, date range definition, and data splitting.



```python
try:
    # 1. Rename the index of df to 'ds' and the sales_column to 'y' to create prophet_df
    prophet_df = df.reset_index()
    prophet_df = prophet_df.rename(columns={df.index.name: 'ds', sales_column: 'y'})

    # Ensure 'ds' is a datetime object
    prophet_df['ds'] = pd.to_datetime(prophet_df['ds'])

    print("DataFrame prepared for Prophet with 'ds' and 'y' columns:")
    display(prophet_df.head())

    # 2. Define the training start date and training end date
    train_start_date = '2020-01-01'
    train_end_date = '2023-12-31'

    # 3. Define the testing start date and testing end date
    test_start_date = '2024-01-01'
    test_end_date = '2024-12-31'

    # 4. Create a training DataFrame, train_df
    train_df = prophet_df[(prophet_df['ds'] >= train_start_date) & (prophet_df['ds'] <= train_end_date)]

    # 5. Create a testing DataFrame, test_df
    test_df = prophet_df[(prophet_df['ds'] >= test_start_date) & (prophet_df['ds'] <= test_end_date)]

    print(f"\nTraining data range: {train_df['ds'].min()} to {train_df['ds'].max()} (Rows: {len(train_df)})")
    print(f"Testing data range: {test_df['ds'].min()} to {test_df['ds'].max()} (Rows: {len(test_df)})")

except NameError:
    print("DataFrame 'df' or 'sales_column' not found. Please ensure previous steps were executed.")
except Exception as e:
    print(f"An error occurred: {e}")
```

### Subtask:
Initialize and train the Prophet model using the training data (2020-2023).


**Reasoning**:
The subtask requires initializing and training the Prophet model. This code block will import the Prophet class, instantiate it with the specified seasonality settings, and then fit the model using the `train_df`.



```python
from prophet import Prophet

try:
    # Initialize the Prophet model
    model = Prophet(
        yearly_seasonality=True,
        weekly_seasonality=False,
        daily_seasonality=False
    )

    # Fit the model to the training data
    model.fit(train_df)

    print("Prophet model initialized and trained successfully on train_df.")

except NameError:
    print("DataFrame 'train_df' not found. Please ensure previous steps were executed.")
except Exception as e:
    print(f"An error occurred during Prophet model initialization or training: {e}")
```

## Calculate and Display Evaluation Metrics

### Subtask:
Calculate the Mean Absolute Error (MAE), Root Mean Squared Error (RMSE), and Mean Absolute Percentage Error (MAPE) between the Prophet model's 2024 predictions and the actual 2024 sales. Present these metrics in a table below the visualization.


**Reasoning**:
To evaluate the model's performance, I need to calculate the Mean Absolute Error (MAE), Root Mean Squared Error (RMSE), and Mean Absolute Percentage Error (MAPE) between the actual 2024 sales and the predicted 2024 sales. I will import the necessary functions, perform the calculations, and then display them in a structured DataFrame.



```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

try:
    # Ensure actual and predicted values are aligned for 2024
    # test_df contains actuals for 2024
    # forecast_2024_filtered contains predictions for 2024
    actual_2024 = test_df['y']
    predicted_2024 = forecast_2024_filtered['yhat']

    # Aligning the dataframes by 'ds' to ensure correct comparison
    # This step is crucial if the dataframes weren't perfectly aligned by index previously.
    # However, given the generation of test_df and forecast_2024_filtered, they should already be aligned.
    # Let's perform a merge to be safe and explicit.
    comparison_df = pd.merge(
        test_df[['ds', 'y']],
        forecast_2024_filtered[['ds', 'yhat']],
        on='ds',
        how='inner'
    )

    actual_2024_aligned = comparison_df['y']
    predicted_2024_aligned = comparison_df['yhat']

    # 1. Calculate Mean Absolute Error (MAE)
    mae = mean_absolute_error(actual_2024_aligned, predicted_2024_aligned)

    # 2. Calculate Root Mean Squared Error (RMSE)
    rmse = np.sqrt(mean_squared_error(actual_2024_aligned, predicted_2024_aligned))

    # 3. Calculate Mean Absolute Percentage Error (MAPE)
    # Avoid division by zero by replacing 0 actual values with a small number if necessary
    # For monthly sales, 0 is unlikely but good practice to consider
    mape = np.mean(np.abs((actual_2024_aligned - predicted_2024_aligned) / actual_2024_aligned)) * 100

    # Create a DataFrame to display the metrics
    metrics_df = pd.DataFrame({
        'Metric': ['MAE', 'RMSE', 'MAPE'],
        'Value': [mae, rmse, mape]
    })

    print("\nEvaluation Metrics for 2024 Predictions:")
    display(metrics_df)

except NameError:
    print("One or more required DataFrames (test_df, forecast_2024_filtered) not found. Please ensure previous steps were executed.")
except Exception as e:
    print(f"An error occurred during metric calculation: {e}")
```

## Prepare Data for SARIMA and Split

### Subtask:
Prepare the DataFrame in a suitable format for SARIMA. Then, split the data into a training set (January 2020 - December 2023) and a testing set (January 2024 - December 2024).


**Reasoning**:
The subtask requires extracting the 'Sales' column into a Series, defining training and testing date ranges, splitting the data accordingly, and then verifying the split by printing the date ranges and number of observations for both sets.



```python
try:
    # 1. Extract the 'Sales' column to create a new pandas Series named sales_series.
    # df already has DatetimeIndex and sales_column is identified as 'Sales'.
    sales_series = df[sales_column]

    # 2. Define the start and end dates for the training period
    train_start_date_sarima = '2020-01-01'
    train_end_date_sarima = '2023-12-31'

    # 3. Define the start and end dates for the testing period
    test_start_date_sarima = '2024-01-01'
    test_end_date_sarima = '2024-12-31'

    # 4. Create the training data Series, train_data_sarima
    train_data_sarima = sales_series.loc[train_start_date_sarima:train_end_date_sarima]

    # 5. Create the testing data Series, test_data_sarima
    test_data_sarima = sales_series.loc[test_start_date_sarima:test_end_date_sarima]

    # 6. Print the start and end dates and the number of observations
    print("SARIMA Data Split Verification:")
    print(f"Training data range: {train_data_sarima.index.min().strftime('%Y-%m-%d')} to {train_data_sarima.index.max().strftime('%Y-%m-%d')} (Observations: {len(train_data_sarima)})")
    print(f"Testing data range: {test_data_sarima.index.min().strftime('%Y-%m-%d')} to {test_data_sarima.index.max().strftime('%Y-%m-%d')} (Observations: {len(test_data_sarima)})")

except NameError:
    print("DataFrame 'df' or 'sales_column' not found. Please ensure previous steps were executed.")
except Exception as e:
    print(f"An error occurred during SARIMA data preparation or splitting: {e}")
```

## Determine SARIMA Parameters and Train Model

### Subtask:
Automatically determine the optimal SARIMA parameters (p, d, q)(P, D, Q)s using `auto_arima` from `pmdarima` and then train the SARIMA model using the training data (2020-2023).


**Reasoning**:
The subtask requires automatically determining optimal SARIMA parameters and training the model using `auto_arima` from `pmdarima`, then printing a summary. This code block will perform these actions.



```python
from pmdarima import auto_arima

try:
    # Automatically determine the optimal SARIMA parameters and fit the model
    sarima_model = auto_arima(train_data_sarima,
                              start_p=1, start_q=1,
                              test='adf',           # Use ADF test to find optimal 'd'
                              max_p=3, max_q=3,
                              m=12,                 # Frequency of the time series (monthly data)
                              start_P=0, start_Q=0,
                              max_P=2, max_Q=2,
                              seasonal=True,        # Enable seasonal ARIMA
                              d=None, D=None,       # Let auto_arima determine 'd' and 'D'
                              trace=False,          # Do not print results at each step
                              error_action='ignore',
                              suppress_warnings=True,
                              stepwise=True         # Use stepwise search to speed up
                             )

    print("SARIMA model fitted successfully. Optimal parameters found.")

    # Print a summary of the fitted model
    print("\nSARIMA Model Summary:")
    print(sarima_model.summary())

except NameError:
    print("DataFrame 'train_data_sarima' not found. Please ensure previous steps were executed.")
except Exception as e:
    print(f"An error occurred during SARIMA model fitting: {e}")
```

**Reasoning**:
The subtask requires generating predictions for the year 2024 using the trained SARIMA model. This involves using the `predict` method of the `sarima_model` to forecast the next 12 steps (for 2024) and capture the prediction intervals.



```python
try:
    # Generate predictions for the next 12 steps (representing 2024)
    # The get_forecast method is used to obtain predictions along with confidence intervals
    n_periods = len(test_data_sarima) # Number of observations in the test set (12 for 2024)
    forecast_sarima_2024 = sarima_model.predict(n_periods=n_periods, return_conf_int=True, alpha=0.05)

    # The prediction result is a tuple (predictions, conf_int)
    predictions_sarima_2024 = forecast_sarima_2024[0]
    conf_int_sarima_2024 = forecast_sarima_2024[1]

    # Create a DataFrame for SARIMA predictions
    sarima_predictions_df = pd.DataFrame({
        'ds': test_data_sarima.index, # Use the index of the test set for dates
        'yhat': predictions_sarima_2024,
        'yhat_lower': conf_int_sarima_2024[:, 0],
        'yhat_upper': conf_int_sarima_2024[:, 1]
    })

    print("SARIMA predictions for 2024 generated successfully:")
    display(sarima_predictions_df.head())
    display(sarima_predictions_df.tail())

except NameError:
    print("SARIMA model 'sarima_model' or 'test_data_sarima' not found. Please ensure previous steps were executed.")
except Exception as e:
    print(f"An error occurred during SARIMA prediction generation: {e}")
```
